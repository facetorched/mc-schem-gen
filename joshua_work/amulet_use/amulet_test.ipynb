```python
import amulet
from amulet.api.level import World
from amulet.api.selection import SelectionBox
from amulet.api.data_types import Dimension
from mcschematic_plus import MCSchematicPlus, read_tiff, read_mesh, read_npy
from mcschematic import Version
import numpy as np

print("successfully imported!")

# --- Paths ---
world_path = r"C:\Users\Public\PrismLauncher\instances\26.2\minecraft\saves\Cellgen World"
input_path = r"C:\Users\public.QCB-MC-VR\Documents\GitHub\Kevin File converter\mcschematic-plus\joshua_work\cell_input\martini2_cell.npy"
CELL_NAME = "amulet_output"
schem_path = f"joshua_work/output/{CELL_NAME}"


# --- Create Schematic
schem = MCSchematicPlus()
colors = {
    0 : "minecraft:lime_stained_glass",
    1 : "minecraft:blue_concrete",
    2 : "minecraft:red_concrete",
    3 : "minecraft:yellow_concrete",
    4 : "minecraft:cyan_concrete",
    5 : "minecraft:purple_concrete"
}

cell_data = np.load(input_path)
cell_data = cell_data[0]

for i, component in enumerate(colors.keys()):
    vol = cell_data[..., i] # already bool
    # mask any previously placed components to avoid overlap
    for j in range(i):
        vol = vol & ~cell_data[..., j]
    schem.placeVolume(vol, colors[component])

schem.save(schem_path)

print("did schem")

# --- Load world ---
world = amulet.load_level(world_path)

# --- Load schematic ---
structure = world.load_structure(schem_path)

# --- Target coordinates ---
# Location is where the center of the structure will be... when pasted
x, y, z = 0, -60, 0

# --- Create a selection box the size of the structure ---
width, height, length = structure.bounds.size
selection = SelectionBox((x, y, z), (x + width, y + height, z + length))

# --- Paste into the overworld ---
world.paste(
    dimension=Dimension.OVERWORLD,
    location=(x, y, z),
    structure=structure
)

# --- Save and close ---
world.save()
world.close()

print("Structure placed successfully.")
```
